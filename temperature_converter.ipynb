```python
num1 = float(input("Enter a Value to convert from Celsius to Fahrenheit:"))
num2 = float(input("Enter a Value to convert from Fahrenheit to Celsius"))

Celsius = num1 * 9/5 + 32
Fahrenheit= num2 - 32 * 5/9

print("")
print("----------RESULT----------")
print("Celsius to Fahrenheit:", Fahrenheit,"°C")
print("Fahrenheit to Celsius:", Celsius,"℉")
```
